## 3. Methodology

*Assumptions, features, label definition, baseline, validation design, leakage checks.*

```python
from IPython.display import Markdown, display

display(Markdown("""
# 3. Methodology

## Features

The analysis uses observed search and engagement signals:

- `gsc_impressions`
- `gsc_clicks`
- `gsc_avg_position`
- `ga4_sessions`
- `ga4_engaged_sessions`

Identifiers such as `client_hash_id` and `content_hash_id` are not used as
model features.

## Baseline

Before modelling, a rule-based action score was created in Week 4 using the
available search and engagement signals.

The baseline provides a reference point for evaluating the later model rather
than assuming that a machine-learning model is automatically better.

## Model

A Random Forest regression model was trained using the same core signals as
the baseline.

## Validation

The model was evaluated using a **time-aware split**. Earlier observations
were used for training and later observations were held out for testing.

This design better reflects the real decision setting because information
from the future should not be available when making a decision about a later
observation.

## Target / label

An important limitation is that the current experiment does **not** use a
verified future business outcome as its target.

Instead, the model was evaluated against the Week-4 baseline score.

Therefore, the measured result indicates agreement with the baseline scoring
rule rather than proven improvement in future content performance.

## Leakage checks

The feature set was audited to exclude:

- identifiers,
- report date as a model feature,
- explicit future-window fields,
- label-derived fields.

`report_date` was used only to construct the time-aware validation split.

## Evaluation

The primary comparison metric was **mean absolute error (MAE)** between the
model prediction and the baseline score.

The result should therefore be interpreted as a measure of agreement with the
baseline, not as evidence of causal improvement.
"""))
```

## 5. Limitations

*What this work cannot claim.*

```python
from IPython.display import Markdown, display

display(Markdown("""
# 5. Limitations & Honest Framing

## Key limitations

### 1. The target is the baseline score

The current Random Forest experiment predicts the Week-4 rule-based baseline
score. It does not predict a verified future content-performance outcome.

Therefore, the measured MAE shows agreement with the baseline rather than
demonstrating that the model improves future decisions.

### 2. Strong dependence on search volume

Approximately 98.8% of the model's feature importance is attributed to
`gsc_impressions`.

This suggests that the model is largely reproducing a volume-driven ranking
pattern rather than demonstrating strong independent predictive value from
the other signals.

### 3. Observational data

The analysis uses observed search and engagement measurements. Associations
between these signals and content performance should not be interpreted as
causal effects.

### 4. Limited observation window

The working analysis covers June 2026. A longer time window would provide
more evidence about seasonality, changing search behaviour, and stability
over time.

### 5. Human review remains necessary

The resulting ranking is intended for decision support. It should not
automatically trigger content edits, publication, deletion, redirects, or
other high-impact actions.

## Honest claim

The safest conclusion from this analysis is:

> The measured signals can support a directional, human-reviewed content
> prioritization workflow, and the Random Forest showed close agreement with
> the existing baseline under a time-aware evaluation. The current evidence
> does not establish that the model improves future content performance.

This framing keeps the conclusions within the evidence produced by the
experiment.
"""))
```

## 6. Ranked recommendations

*The action playbook output — the paper's recommendations section.*

```python
from IPython.display import Markdown, display

display(Markdown("""
# 6. Ranked Recommendations

The ranked queue is intended to help a human reviewer decide where to look
first.

## Recommended action hierarchy

### 1. Optimize Existing Content
**Reason:** GOOD_POSITION

Pages already achieving strong search positions may have an opportunity for
careful optimization rather than a complete rewrite.

**Human check:** confirm that the page satisfies search intent and that there
is a meaningful content or conversion improvement available.

### 2. Review and Improve
**Reason:** MID_POSITION

Pages with moderate search positions should be reviewed for content gaps,
search-intent alignment, and opportunities to improve relevance.

**Human check:** determine whether the page has a realistic opportunity to
move higher before investing effort.

### 3. Investigate Content
**Reason:** LOW_VISIBILITY

Pages with weaker visibility signals should be investigated before deciding
whether a refresh is appropriate.

**Human check:** distinguish genuine content problems from low search demand,
technical issues, seasonality, or other external factors.

### 4. Manual Review
**Reason:** INSUFFICIENT_SIGNAL

Pages without enough reliable signal should not receive an automated action.

**Human check:** gather additional context before deciding what to do.

## Refresh / decay insight

The broader FlyRank research context indicates that content performance can
change with content age and that refreshed older content can show stronger
observed performance.

This should be treated as a **directional observation**, not proof that
refreshing a page will cause improved performance.

## Overall recommendation

Use the ranked queue to prioritize **human attention**, beginning with pages
that have the strongest measurable signals and a clear reason code.

Do not treat rank as an automatic instruction to modify content.
"""))
```
