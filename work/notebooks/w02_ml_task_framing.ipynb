```python
from IPython.display import Markdown, display

display(Markdown("""
## Target or Proxy

**Target:** `needs_refresh`

**Prediction:**

- **1** → Page should be refreshed
- **0** → Page does not need a refresh

**Why this target?**

The ideal target would come from historical decisions or observed improvements after content updates. If those labels are not available, we can create a proxy target using measurable signals such as content age, click-through rate (CTR), impressions, and average search position.

This target allows the model to learn patterns that help prioritize pages for future content refreshes.
"""))
```
