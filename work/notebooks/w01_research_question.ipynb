## 2. The question: decision, action, cost of a wrong call

*What decision does your work improve? Who acts on it? What does a wrong recommendation cost?*

```python
from IPython.display import display, Markdown

display(Markdown("""
## Decision

The decision is **which existing content pages should be prioritized for review
or refresh**.

## Who acts on it?

A content/SEO team can use the ranked recommendations to decide which pages
should be reviewed first.

## Action

For a high-priority page, the team can review its search performance and decide
whether to refresh the content, improve its title and metadata, or leave it
unchanged.

## Cost of a wrong call

A false positive can waste content-team time on a page that did not need a
refresh.

A false negative can cause a potentially valuable page to be overlooked,
delaying an improvement opportunity.

The goal is therefore **decision support**, not an automatic decision.
"""))
```

## 4. Careful words: what I can and can't claim

*Write what your work will be able to say (observed, directional, decision-support) — and what it never will (causal proof, 'predicting Google').*

```python
from IPython.display import display, Markdown

display(Markdown("""
## Careful Words: What I Can and Can't Claim

### What I can claim

This work can describe **observed relationships** between content performance
signals such as impressions, clicks, search position, and engagement.

The results can provide **directional evidence** about which content pages may
deserve attention first.

The final output is **decision-support** for a content team: it helps rank pages
for review or refresh.

### What I cannot claim

This analysis does **not prove causality**. A relationship between a signal and
content performance does not mean that changing that signal will cause the
outcome to change.

It also does not claim to **predict Google** or guarantee future search rankings.

The model should therefore be treated as a prioritization tool rather than an
automatic decision-maker.
"""))
```
